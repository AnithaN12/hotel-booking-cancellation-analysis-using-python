# Data Analysis and Visualizations

```python
cancelled_perc = df['is_canceled'].value_counts(normalize = True)
print(cancelled_perc)

plt.figure(figsize=(5,4))
plt.title('Reservation Status Count')
bars = plt.bar(['Not canceled','canceled'],df['is_canceled'].value_counts(), edgecolor='k', width=0.7)
plt.bar_label(bars)
plt.show()
          
```

```python
plt.figure(figsize = (8,4))
ax1 = sns.countplot(x='hotel', hue='is_canceled', data=df, edgecolor='k', palette= 'Blues')
ax1.legend(bbox_to_anchor=(1,1))
plt.title('Reservation status in different hotels', size = 20)
plt.xlabel=('Hotel')
plt.ylabel=('Number of Reservations')
plt.legend(['Not canceled', 'Canceled'])
for container in ax1.containers:
    ax1.bar_label(container)
plt.show()
```

```python
resort_hotel=df[df['hotel']=='Resort Hotel']
resort_hotel['is_canceled'].value_counts(normalize=True)
```

```python
city_hotel=df[df['hotel']=='City Hotel']
city_hotel['is_canceled'].value_counts(normalize=True)
```

```python
resort_hotel = resort_hotel.groupby('reservation_status_date')[['adr']].mean()
city_hotel = city_hotel.groupby('reservation_status_date')[['adr']].mean()
```

```python
plt.figure(figsize= (20,8))
plt.title('Average Daily Rate in City and Resort Hotel', fontsize =30)
plt.plot(resort_hotel.index, resort_hotel['adr'], label = 'Resort Hotel')
plt.plot(city_hotel.index, city_hotel['adr'], label = 'City Hotel')
plt.legend(fontsize = 20)
plt.show()
```

```python
df['month'] = df['reservation_status_date'].dt.month_name()
month_order=['January', 'February', 'March','April', 'May','June', 'July', 'August', 'September','October','November','December']
plt.figure(figsize=(16,7))
ax1=sns.countplot(x='month',hue='is_canceled',data=df,order=month_order, edgecolor='k', palette='bright')
plt.title('Reservation Status per Month',size=20)
plt.legend(['Not canceled', 'Canceled'],bbox_to_anchor=(1,1))
ax1.set_xlabel('Month')
ax1.set_ylabel('Number of Reservations')
for container in ax1.containers:
    ax1.bar_label(container)
plt.show()
```

```python
Chart_data = df[df['is_canceled']==1].groupby('month')[['adr']].sum().reset_index()
month_order=['January', 'February', 'March','April', 'May','June', 'July', 'August', 'September','October','November','December']
plt.figure(figsize= (15,7))
plt.title('ADR per Month', fontsize = 20)
ax1 = sns.barplot(x='month',y='adr',data = Chart_data,order = month_order, edgecolor='k')
for container in ax1.containers:
    ax1.bar_label(container)
plt.show()
```

```python
canceled_data = df[df['is_canceled']==1]
top_10_country = canceled_data['country'].value_counts()[:10]
plt.figure(figsize=(7,7))
plt.title('Top 10 Countries with Reservation Canceled')
plt.pie(top_10_country,autopct='%.2f', labels= top_10_country.index)
plt.show()
```

```python
canceled_df_adr = canceled_data.groupby('reservation_status_date')[['adr']].mean()
canceled_df_adr.reset_index(inplace = True)
canceled_df_adr.sort_values('reservation_status_date',inplace = True)

not_canceled_data = df[df['is_canceled']==0]
not_canceled_df_adr = not_canceled_data.groupby('reservation_status_date')[['adr']].mean()
not_canceled_df_adr.reset_index(inplace = True)
not_canceled_df_adr.sort_values('reservation_status_date',inplace = True)

plt.figure(figsize=(20,6))
plt.title('Average Daily Rate',size=20)
plt.plot(not_canceled_df_adr['reservation_status_date'],not_canceled_df_adr['adr'], label='not_canceled')
plt.plot(canceled_df_adr['reservation_status_date'],canceled_df_adr['adr'], label='canceled')                                       
plt.legend()
plt.show()
```
